Use Statevector and Operator classes to explore the behaviour of multiple qubits.

```python
from qiskit import __version__
from qiskit.quantum_info import Statevector, Operator
from numpy import sqrt
print(__version__)
```

```python
#tensor product of two vectors psi = ket(0) x ket(1)
zero = Statevector.from_label("0")
one  = Statevector.from_label("1")
psi = zero.tensor(one)
display(psi.draw("latex"))
```

```python
#Bell's state phi_plus
phi_plus = (Statevector.from_label("00") + Statevector.from_label("11")) / sqrt(2)
display(phi_plus.draw("latex"))
```

```python
#Bell's state psi_minus
psi_minus = (Statevector.from_label("01") - Statevector.from_label("10")) / sqrt(2)
display(psi_minus.draw("latex"))
```

```python
#1. Find tensor product of Bell's states phi_plus and psi_minus.
#2. Check if psi is a valid state.
psi = phi_plus.tensor(psi_minus)
display(psi.draw("latex"))
print(f"psi is a valid state?", psi.is_valid())
```

```python
#tensor product of two operators
H = Operator.from_label("H") #Hadamard
I = Operator.from_label("I") #Identity
X = Operator.from_label("X") #Not

HIX = H.tensor(I).tensor(X)
display(HIX.draw("latex"))
print(f"dimensions of HIX:", HIX.dim)
```

```python
#Construct Toffoli operation
Zero = Operator.from_label("0")
One  = Operator.from_label("1")
I    = Operator.from_label("I")
X    = Operator.from_label("X")

Tfl = (
    (Zero ^ I ^ I)
    +
    (One ^ ((Zero ^ I) + (One ^ X)))
)

display(Tfl.draw("latex"))
print(f"Dimensions of Toffoli operation on 3 qubits:", Tfl.dim)
```
