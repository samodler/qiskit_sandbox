```python
from qiskit import QuantumCircuit as QC
from qiskit.quantum_info import Statevector, Operator
import numpy as np

sqrt = np.sqrt
```

# Oct 5

```python
#CX gate
CX = (
    Operator(np.array([[1, 0], [0, 0]])) ^ Operator.from_label('I')
    ) + (
        Operator(np.array([[0, 0], [0, 1]])) ^ Operator.from_label('X')
    )
display(CX.draw("latex"))
```

```python
# 1. HSHT circuit
# 2.Apply to |0>
H = Operator(np.sqrt(2)/2 * np.array([[1, 1], [1, -1]]))
S = Operator(np.array([[1, 0], [0, 1j]])) #phase rotation by pi/2 rad
T = Operator(np.array([[1, 0], [0, (1+1j)/np.sqrt(2)]])) #phase rotation by pi/4 rad

HSHT = T @ H @ S @ H
zero = Statevector(np.array([1, 0]))
```

```python
updated_state = zero.evolve(HSHT)
display(updated_state.draw("latex"))
```

```python
I = Operator(np.array([[1, 0], [0, 1]]))
X = Operator(np.array([[0, 1], [1, 0]]))
Zero = Operator.from_label("0")
One = Operator.from_label("1")
```

```python
Gate = (I ^ Zero) + (X ^ One)
display(Gate.draw("latex"))
```

```python
phi_plus = Statevector(np.array([sqrt(1/2), 0, 0, sqrt(1/2)]))
display(phi_plus.draw("latex"))
```

```python
result, state = phi_plus.measure()
print(f"Measured: {result}\nState after measurement:")
display(state.draw("latex"))
```
