```python
from qiskit import QuantumCircuit as QC
from qiskit.quantum_info import Statevector,  Operator
from qiskit.visualization import array_to_latex as atx
```

### 2 qubits circuit

```python
circ = QC(2)
circ.h(1)
circ.draw('mpl')
```

```python
display(Statevector(circ).draw("latex"))
```

### 4 qubits circuit

```python
circ = QC(4)
circ.cx(0, 1)
circ.cx(0, 2)
circ.h(0)
circ.cx(1, 3)
circ.draw()
```

```python
display(Statevector(circ).draw("latex"))
```

```python
atx(Operator(circ))
```

### Function that generate GHZ circuit/state for n qubits
[Wikipedia - GHZ state](https://en.wikipedia.org/wiki/Greenberger%E2%80%93Horne%E2%80%93Zeilinger_state)

```python
def ghz(n):
    circ = QC(n)
    circ.h(0)
    for i in range(1, n):
        circ.cx(i-1, i)
    return circ
```

```python
n = 5
circ = ghz(n)
circ.draw("mpl")
```

```python
ghz_state = Statevector(circ)
display(ghz_state.draw("latex"))
```
