```python
from qiskit.quantum_info import Statevector,  Operator
import numpy as np
```

### Real vectors

```python
psi = Statevector(np.array([np.sqrt(2)/2, np.sqrt(2)/2]))
display(psi.draw('latex'))
```

```python
phi = Statevector(np.array([1/2, np.sqrt(3)/2]))
display(phi.draw('latex'))
```

```python
dotProduct = psi.inner(phi)
print(f'<psi|phi> = {dotProduct}')
```

### Complex vectors

```python
psi = Statevector(np.array([1j/np.sqrt(2),1/np.sqrt(2)]))
display(psi.draw('latex'))
```

```python
psi.is_valid()
```

```python
phi = Statevector(np.array([np.sqrt(3)/2, 1j/2]))
display(phi.draw('latex'))
```

```python
phi.is_valid()
```

```python
psiPhi = psi.inner(phi)
print(f'<psi|phi> = {psiPhi}')
```

```python
phiPsi = phi.inner(psi)
print(f'<phi|psi> = {phiPsi}')
```
